##### | Testing the Multi-Agent System

```python
from graph import graph

def run_test(thread_id: str, user_input: str) -> dict:
    run_config1 = {
        "configurable": {
            "thread_id": thread_id,
        }
    }

    return graph.invoke(
        {
            "user_input": user_input,
            "route": "",
            "output": "",
            "messages": [],
        },
        config=run_config1,
    )


# Test 1: SEO Blog Writer
seo_result = run_test(
    "test-seo",
    "Write a long-form SEO blog post about the benefits of renewable energy.",
)

print("SEO route:", seo_result["route"])
print(seo_result["output"])


# Test 2: X/Twitter Writer
x_result = run_test(
    "test-x",
    "Write an engaging X post about artificial intelligence trends.",
)

print("X route:", x_result["route"])
print(x_result["output"])


# Test 3: General Handler
general_result = run_test(
    "test-general",
    "Hello, how are you?",
)

print("General route:", general_result["route"])
print(general_result["output"])
```

```python
assert seo_result["route"] == "seo_blog_writer"
assert x_result["route"] == "x_blog_writer"
assert general_result["route"] == "general"

print("All three routes passed.")
```
